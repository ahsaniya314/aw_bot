## 1. Definisikan Ground Truth dan Fungsi Perhitungan CER/WER

### Rumus Perhitungan CER (Character Error Rate)
CER mengukur kesalahan pada tingkat karakter. Rumusnya adalah:
$$CER = \frac{S + D + I}{N}$$

**Keterangan:**
* **S (Substitution):** Jumlah karakter yang salah dikenali (digantikan oleh karakter lain).
* **D (Deletion):** Jumlah karakter yang hilang (tidak terdeteksi).
* **I (Insertion):** Jumlah karakter tambahan yang disisipkan (tidak ada di ground truth).
* **N:** Jumlah total karakter pada teks referensi (ground truth).

**Akurasi Karakter (Character Accuracy):**
$$\text{Akurasi CER} = 1 - CER = \frac{N - (S + D + I)}{N}$$
*(Catatan: Jika error rate lebih besar dari 100%, akurasi dibatasi minimal 0%)*

---

### Rumus Perhitungan WER (Word Error Rate)
WER mengukur kesalahan pada tingkat kata. Rumusnya adalah:
$$WER = \frac{S + D + I}{N}$$

**Keterangan:**
* **S (Substitution):** Jumlah kata yang salah dikenali (digantikan oleh kata lain).
* **D (Deletion):** Jumlah kata yang hilang.
* **I (Insertion):** Jumlah kata tambahan yang disisipkan.
* **N:** Jumlah total kata pada teks referensi (ground truth).

**Akurasi Kata (Word Accuracy):**
$$\text{Akurasi WER} = 1 - WER = \frac{N - (S + D + I)}{N}$$
*(Catatan: Jika error rate lebih besar dari 100%, akurasi dibatasi minimal 0%)*

```python
ground_truth = """28 - 11 - 2029

Nama sales = Budi Santoso

Ambil barang
lolipop 100 ctn
meses 20 ctn
miksu 50 ctn

30 - 11 - 2029

Nama = Budi Santoso

Bayar cicilan : Rp 73.600.000

02 - 12 - 2029

Nama = Budi Santoso

Bayar cicilan : Rp 10.000.000"""

def calculate_alignment(ref, hyp):
    """Calculates Levenshtein distance alignment returning S, D, I, N, error_rate, and accuracy."""
    n_ref = len(ref)
    n_hyp = len(hyp)
    
    # Create distance matrix
    d = [[0] * (n_hyp + 1) for _ in range(n_ref + 1)]
    for i in range(n_ref + 1):
        d[i][0] = i
    for j in range(n_hyp + 1):
        d[0][j] = j
        
    for i in range(1, n_ref + 1):
        for j in range(1, n_hyp + 1):
            cost = 0 if ref[i-1] == hyp[j-1] else 1
            d[i][j] = min(
                d[i-1][j] + 1,        # deletion
                d[i][j-1] + 1,        # insertion
                d[i-1][j-1] + cost    # substitution
            )
            
    # Backtracking to find S, D, I
    S, D, I = 0, 0, 0
    i, j = n_ref, n_hyp
    while i > 0 or j > 0:
        if i > 0 and j > 0:
            cost = 0 if ref[i-1] == hyp[j-1] else 1
            if d[i][j] == d[i-1][j-1] + cost:
                if cost == 1:
                    S += 1
                i -= 1
                j -= 1
            elif d[i][j] == d[i-1][j] + 1:
                D += 1
                i -= 1
            elif d[i][j] == d[i][j-1] + 1:
                I += 1
                j -= 1
            else:
                i -= 1
                j -= 1
        elif i > 0:
            D += 1
            i -= 1
        elif j > 0:
            I += 1
            j -= 1
            
    err_rate = d[n_ref][n_hyp] / n_ref if n_ref > 0 else (1.0 if n_hyp > 0 else 0.0)
    acc = max(0.0, 1.0 - err_rate)
    return S, D, I, n_ref, err_rate, acc

def calculate_cer(reference, hypothesis):
    """Calculate Character Error Rate (CER)."""
    _, _, _, _, cer, _ = calculate_alignment(list(reference), list(hypothesis))
    return cer

def calculate_wer(reference, hypothesis):
    """Calculate Word Error Rate (WER)."""
    _, _, _, _, wer, _ = calculate_alignment(reference.split(), hypothesis.split())
    return wer

print("✅ Ground truth, fungsi pencocokan Levenshtein (S, D, I, N), dan fungsi CER/WER siap!")
```

## 3. Fungsi untuk Memproses Satu Gambar

```python
def process_single_image(img_number):
    """Memproses satu gambar dan menyimpan hasilnya ke file JSON."""
    img_filename = f"{img_number}.jpg"
    tes_gambar_dir = project_root / "tes gambar"
    img_path = str(tes_gambar_dir / img_filename)
    
    if not os.path.exists(img_path):
        print(f"❌ Error: Gambar {img_filename} tidak ditemukan di {tes_gambar_dir}")
        return False
    
    print(f"\nMemproses gambar {img_filename}...")
    print("=" * 80)
    
    # Ekstrak teks
    extracted_text = ocr_service.extract_text(img_path)
    
    # Hitung detail CER
    cer_s, cer_d, cer_i, cer_n, cer, cer_acc = calculate_alignment(list(ground_truth), list(extracted_text))
    # Hitung detail WER
    wer_s, wer_d, wer_i, wer_n, wer, wer_acc = calculate_alignment(ground_truth.split(), extracted_text.split())
    
    print(f"CER: {cer:.2%} (S={cer_s}, D={cer_d}, I={cer_i}, N={cer_n}) -> Akurasi: {cer_acc:.2%}")
    print(f"WER: {wer:.2%} (S={wer_s}, D={wer_d}, I={wer_i}, N={wer_n}) -> Akurasi: {wer_acc:.2%}")
    print(f"\nTeks diekstrak:\n{extracted_text}")
    print("-" * 80)
    
    # Simpan hasil ke file JSON
    results_dir = project_root / "ocr_results"
    results_dir.mkdir(exist_ok=True)
    result_file = results_dir / f"result_{img_number}.json"
    
    result_data = {
        "filename": img_filename,
        "cer": cer,
        "cer_details": {
            "S": cer_s,
            "D": cer_d,
            "I": cer_i,
            "N": cer_n,
            "accuracy": cer_acc
        },
        "wer": wer,
        "wer_details": {
            "S": wer_s,
            "D": wer_d,
            "I": wer_i,
            "N": wer_n,
            "accuracy": wer_acc
        },
        "extracted_text": extracted_text
    }
    
    with open(result_file, "w", encoding="utf-8") as f:
        json.dump(result_data, f, ensure_ascii=False, indent=2)
    
    print(f"\n✅ Hasil disimpan ke: {result_file}")
    return True

print("✅ Fungsi process_single_image siap!")
```

## 4. Proses Gambar Satu Per Satu (Dengan Delay Opsional)

Jalankan sel berikut untuk memproses gambar satu per satu. Setiap selesai 4 gambar, akan ada delay 1 menit untuk menghindari rate limit!

```python
def process_images_with_delay(start_num, end_num, delay_after=4, delay_seconds=60):
    """Memproses gambar dengan delay otomatis."""
    count = 0
    for img_num in range(start_num, end_num + 1):
        process_single_image(img_num)
        count += 1
        
        # Delay setelah N gambar
        if count % delay_after == 0 and img_num != end_num:
            print("\n" + "=" * 100)
            print(f"✅ Sudah memproses {count} gambar. Delay {delay_seconds} detik untuk menghindari rate limit...".center(100))
            print("=" * 100)
            time.sleep(delay_seconds)

# Contoh: Proses gambar 6-9 dengan delay 1 menit setiap 4 gambar
process_images_with_delay(6, 9, delay_after=4, delay_seconds=60)
```

## 5. Agregasi Semua Hasil dan Tarik Kesimpulan

```python
def aggregate_results():
    """Menggabungkan semua hasil dan menampilkan ringkasan detail perhitungan."""
    results_dir = project_root / "ocr_results"
    
    if not results_dir.exists():
        print(f"❌ Error: Folder {results_dir} tidak ditemukan! Jalankan terlebih dahulu process_single_image untuk setiap gambar.")
        return
    
    # Baca semua file result
    all_results = []
    for i in range(1, 10):
        result_file = results_dir / f"result_{i}.json"
        if result_file.exists():
            with open(result_file, "r", encoding="utf-8") as f:
                data = json.load(f)
                all_results.append(data)
                print(f"[OK] Memuat hasil untuk {data['filename']}")
        else:
            print(f"[WARN] Hasil untuk {i}.jpg belum ditemukan!")
    
    if not all_results:
        print("\n❌ Tidak ada hasil yang bisa diolah!")
        return
    
    # Hitung total dan rata-rata
    total_cer = sum(r["cer"] for r in all_results)
    total_wer = sum(r["wer"] for r in all_results)
    avg_cer = total_cer / len(all_results)
    avg_wer = total_wer / len(all_results)
    
    # Cari performa terbaik dan terburuk
    best_cer = min(all_results, key=lambda x: x["cer"])
    worst_cer = max(all_results, key=lambda x: x["cer"])
    best_wer = min(all_results, key=lambda x: x["wer"])
    worst_wer = max(all_results, key=lambda x: x["wer"])
    
    # Tampilkan ringkasan
    print("\n" + "=" * 125)
    print("RINGKASAN AKHIR PENGUJIAN OCR MISTRAL (9 GAMBAR)".center(125))
    print("=" * 125)
    print(f"\n1. JUMLAH GAMBAR YANG DIUJI: {len(all_results)} dari 9")
    print("\n2. DETAIL HASIL DAN PERHITUNGAN AKURASI PER GAMBAR:")
    print("-" * 125)
    print(f"{'NAMA FILE':<12} | {'CER':<8} {'CER DETAILS (S/D/I/N)':<24} {'AKURASI CER':<12} | {'WER':<8} {'WER DETAILS (S/D/I/N)':<24} {'AKURASI WER':<12}")
    print("-" * 125)
    for r in all_results:
        # Gunakan fallback jika key detail tidak ditemukan
        cer_det = r.get("cer_details", {"S": "?", "D": "?", "I": "?", "N": "?", "accuracy": max(0.0, 1.0 - r["cer"])})
        wer_det = r.get("wer_details", {"S": "?", "D": "?", "I": "?", "N": "?", "accuracy": max(0.0, 1.0 - r["wer"])})
        
        cer_det_str = f"S={cer_det['S']}, D={cer_det['D']}, I={cer_det['I']}, N={cer_det['N']}"
        wer_det_str = f"S={wer_det['S']}, D={wer_det['D']}, I={wer_det['I']}, N={wer_det['N']}"
        
        print(f"{r['filename']:<12} | {r['cer']:>6.2%} {cer_det_str:<24} {cer_det['accuracy']:>11.2%} | {r['wer']:>6.2%} {wer_det_str:<24} {wer_det['accuracy']:>11.2%}")
    print("-" * 125)
    print(f"\n3. RATA-RATA:")
    print(f"   - Rata-rata CER: {avg_cer:.2%} (Rata-rata Akurasi CER: {1-avg_cer:.2%})")
    print(f"   - Rata-rata WER: {avg_wer:.2%} (Rata-rata Akurasi WER: {1-avg_wer:.2%})")
    print("\n4. PERFORMA TERBAIK:")
    print(f"   - CER Terendah (Akurasi Tertinggi): {best_cer['filename']} ({best_cer['cer']:.2%})")
    print(f"   - WER Terendah (Akurasi Tertinggi): {best_wer['filename']} ({best_wer['wer']:.2%})")
    print("\n5. PERFORMA TERBURUK:")
    print(f"   - CER Tertinggi (Akurasi Terendah): {worst_cer['filename']} ({worst_cer['cer']:.2%})")
    print(f"   - WER Tertinggi (Akurasi Terendah): {worst_wer['filename']} ({worst_wer['wer']:.2%})")
    print("\n" + "=" * 125)
    print("KESIMPULAN".center(125))
    print("=" * 125)
    print(f"\nBerdasarkan pengujian {len(all_results)} gambar dari folder 'tes gambar':")
    if avg_cer < 0.30:
        print(f"[SANGAT BAIK] Performa OCR Mistral SANGAT BAIK dengan rata-rata CER di bawah 30% (Akurasi: {1-avg_cer:.2%})")
    elif avg_cer < 0.50:
        print(f"[CUKUP BAIK] Performa OCR Mistral CUKUP BAIK dengan rata-rata CER di bawah 50% (Akurasi: {1-avg_cer:.2%})")
    elif avg_cer < 0.70:
        print(f"[CUKUP] Performa OCR Mistral CUKUP dengan rata-rata CER di bawah 70% (Akurasi: {1-avg_cer:.2%})")
    else:
        print(f"[PERLU DIPERBAIKI] Performa OCR Mistral PERLU DIPERBAIKI dengan rata-rata CER di atas 70% (Akurasi: {1-avg_cer:.2%})")
    print("\n" + "=" * 125)

aggregate_results()
```
